# Phân loại dữ liệu phẳng với một lớp ẩn

Chào mừng bạn đến với bài tập lập trình tuần 3! Đã đến lúc xây dựng mạng nơ-ron đầu tiên của bạn, một mạng có một lớp ẩn. Bạn sẽ nhận thấy sự khác biệt lớn giữa mô hình này và mô hình hồi quy logistic mà bạn đã triển khai trước đó.

Sau khi hoàn thành bài tập này, bạn sẽ có thể:

- Triển khai mạng nơ-ron phân loại 2 lớp với một lớp ẩn duy nhất

- Tính toán hàm mất mát cross-entropy (entropy chéo)
- Triển khai quá trình lan truyền tiến (forward propagation) và lan truyền ngược (backward propagation)
