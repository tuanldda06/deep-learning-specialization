Điều này cho phép một mạng RNN đơn hướng (một chiều) tận dụng thông tin từ quá khứ để xử lý các đầu vào sau đó. **Ký hiệu**:
- Chỉ số trên $[l]$ biểu thị một đối tượng liên quan đến lớp thứ $l$.

- Chỉ số trên $\langle t \rangle$ biểu thị một đối tượng tại bước thời gian thứ $t$.

- Chỉ số dưới $i$ biểu thị phần tử thứ $i$ của một vectơ.

**Ví dụ**:
- $a^{(2)[3]<4>}_5$ biểu thị giá trị kích hoạt của mẫu huấn luyện thứ 2 (2), tại lớp thứ 3 [3], bước thời gian thứ 4 <4>, và là phần tử thứ 5 trong vectơ. #### Yêu cầu tiên quyết

## 1 - Lan truyền tiến cho Mạng Nơ-ron Hồi quy (RNN) cơ bản

Cuối tuần này, bạn sẽ có cơ hội tạo ra âm nhạc bằng cách sử dụng RNN! Mạng RNN cơ bản mà bạn sẽ triển khai có cấu trúc như sau:

Trong ví dụ này, $T_x = T_y$.

<img src="images/RNN.png" style="width:500;height:300px;">
<caption><center><font color='purple'><b>Figure 1</b>: Basic RNN model </center></caption>

### Kích thước của đầu vào $x$

#### Đầu vào với $n_x$ đơn vị
* Đối với một bước thời gian của một mẫu dữ liệu đầu vào, $x^{(i) \langle t \rangle }$ là một vectơ đầu vào một chiều.
* Lấy ví dụ về ngôn ngữ: một ngôn ngữ có vốn từ vựng gồm 5000 từ có thể được mã hóa one-hot thành một vectơ gồm 5000 đơn vị. Khi đó, $x^{(i)\langle t \rangle}$ sẽ có hình dạng (5000,).
* Ký hiệu $n_x$ được sử dụng ở đây để chỉ số lượng đơn vị trong một bước thời gian của một mẫu dữ liệu huấn luyện.

#### Các bước thời gian với độ dài $T_{x}$
* Mạng thần kinh tái phát (RNN) có nhiều bước thời gian, được đánh chỉ số bằng $t$.
* Trong các bài học, bạn đã thấy một ví dụ huấn luyện đơn lẻ $x^{(i)}$ bao gồm nhiều bước thời gian $T_x$. Trong notebook này, $T_{x}$ sẽ biểu thị số lượng bước thời gian của chuỗi dài nhất.

#### Các lô (batch) có kích thước $m$
* Giả sử ta có các mini-batch, mỗi lô gồm 20 mẫu huấn luyện
* Để tận dụng lợi thế của kỹ thuật vector hóa, bạn sẽ xếp chồng 20 cột chứa các mẫu $x^{(i)}$
* Ví dụ, tensor này có kích thước (5000, 20, 10)
* Bạn sẽ dùng $m$ để ký hiệu số lượng mẫu huấn luyện
* Như vậy, kích thước của một mini-batch là $(n_x, m, T_x)$

#### Tensor 3 chiều có kích thước $(n_{x},m,T_{x})$
* Tensor 3 chiều $x$ với kích thước $(n_x,m,T_x)$ đại diện cho dữ liệu đầu vào $x$ được đưa vào mạng RNN.

#### Lấy một lát cắt 2 chiều cho mỗi bước thời gian: $x^{\langle t \rangle}$
* Tại mỗi bước thời gian, bạn sẽ sử dụng một mini-batch (nhóm nhỏ) các mẫu huấn luyện (thay vì chỉ một mẫu duy nhất).
* Do đó, với mỗi bước thời gian $t$, bạn sẽ sử dụng một lát cắt 2 chiều có kích thước $(n_x,m)$.
* Lát cắt 2 chiều này được ký hiệu là $x^{\langle t \rangle}$. Tên biến trong mã nguồn là `xt`.

### Định nghĩa về trạng thái ẩn $a$

* Giá trị kích hoạt $a^{\langle t \rangle}$ được truyền trong mạng RNN từ bước thời gian này sang bước thời gian khác được gọi là "trạng thái ẩn" (hidden state).

### Kích thước của trạng thái ẩn $a$

* Tương tự như tensor đầu vào $x$, trạng thái ẩn của một mẫu huấn luyện đơn lẻ là một vectơ có độ dài $n_{a}$.
* Nếu xét một mini-batch gồm $m$ mẫu huấn luyện, kích thước (shape) của mini-batch đó là $(n_{a}, m)$.
* Khi tính đến chiều thời gian, kích thước của trạng thái ẩn là $(n_{a}, m, T_x)$.
* Bạn sẽ thực hiện vòng lặp qua các bước thời gian với chỉ số $t$ và thao tác trên một lát cắt 2 chiều của tensor 3 chiều này.
* Lát cắt 2 chiều này được ký hiệu là $a^{\langle t \rangle}$.
* Trong mã nguồn, các tên biến được sử dụng là `a_prev` hoặc `a_next`, tùy thuộc vào hàm đang được triển khai.
* Kích thước của lát cắt 2 chiều này là $(n_{a}, m)$.

### Kích thước của dự đoán $\hat{y}$
* Tương tự như các đầu vào và trạng thái ẩn, $\hat{y}$ là một tensor 3 chiều có hình dạng $(n_{y}, m, T_{y})$
* $n_{y}$: số lượng đơn vị trong vectơ biểu diễn dự đoán
* $m$: số lượng mẫu trong một mini-batch
* $T_{y}$: số bước thời gian trong dự đoán
* Với một bước thời gian $t$ cụ thể, lát cắt 2 chiều $\hat{y}^{\langle t \rangle}$ có hình dạng $(n_{y}, m)$
* Trong mã nguồn, các tên biến là:
- `y_pred`: $\hat{y}$
- `yt_pred`: $\hat{y}^{\langle t \rangle}$

Dưới đây là cách triển khai một RNN:

### Các bước:
1. Triển khai các phép tính cần thiết cho một bước thời gian của RNN.
2. Thực hiện vòng lặp qua $T_x$ bước thời gian để xử lý tất cả các đầu vào, từng bước một.

### 1.1 - Ô RNN (RNN Cell)

Bạn có thể hình dung mạng thần kinh tái phát (RNN) như là sự lặp lại việc sử dụng một ô đơn lẻ. Trước tiên, bạn sẽ triển khai các phép tính cho một bước thời gian (time step) duy nhất. Hình dưới đây mô tả các thao tác cho một bước thời gian của một ô RNN:

<img src="images/rnn_step_forward_figure2_v3a.png" style="width:700px;height:300px;">
<caption><center><font color='purple'><b>Hình 2</b>: Ô RNN cơ bản. Nhận đầu vào là $x^{\langle t \rangle}$ (đầu vào hiện tại) và $a^{\langle t - 1\rangle}$ (trạng thái ẩn trước đó chứa thông tin từ quá khứ), đồng thời xuất ra $a^{\langle t \rangle}$ — giá trị này được chuyển tới ô RNN tiếp theo và cũng được dùng để dự đoán $\hat{y}^{\langle t \rangle}$.
</center></caption>

**`RNN cell` so với `RNN_cell_forward`**:
* Lưu ý rằng một ô RNN xuất ra trạng thái ẩn $a^{\langle t \rangle}$. 
* `RNN cell` được thể hiện trong hình dưới dạng ô bên trong với các đường nét liền.
* Hàm mà bạn sẽ triển khai, `rnn_cell_forward`, còn tính toán cả giá trị dự đoán $\hat{y}^{\langle t \rangle}$.
* `RNN_cell_forward` được thể hiện trong hình dưới dạng ô bên ngoài với các đường nét đứt.

<a name='ex-1'></a>
### Bài tập 1 - rnn_cell_forward

Triển khai ô RNN (RNN cell) được mô tả trong Hình 2.

**Hướng dẫn**:
1. Tính trạng thái ẩn (hidden state) với hàm kích hoạt tanh: $a^{\langle t \rangle} = \tanh(W_{aa} a^{\langle t-1 \rangle} + W_{ax} x^{\langle t \rangle} + b_a)$
2. Sử dụng trạng thái ẩn mới $a^{\langle t \rangle}$ để tính giá trị dự đoán $\hat{y}^{\langle t \rangle} = softmax(W_{ya} a^{\langle t \rangle} + b_y)$. (Hàm `softmax` đã được cung cấp sẵn)
3. Lưu trữ $(a^{\langle t \rangle}, a^{\langle t-1 \rangle}, x^{\langle t \rangle}, parameters)$ vào biến `cache`
4. Trả về $a^{\langle t \rangle}$, $\hat{y}^{\langle t \rangle}$ và `cache`

#### Gợi ý bổ sung
* Thông tin thêm về [numpy.tanh](https://numpy.org/devdocs/reference/generated/numpy.tanh.html)
* Trong bài tập này, đã có sẵn hàm `softmax` để bạn sử dụng. Hàm này nằm trong tệp 'rnn_utils.py' và đã được import (nhập) sẵn.
* Đối với phép nhân ma trận, hãy sử dụng [numpy.dot](https://docs.scipy.org/doc/numpy/reference/generated/numpy.dot.html)

````python
def rnn_cell_forward(xt, a_prev, parameters):
    """
    Thực hiện một bước lan truyền tiến (forward step) duy nhất của ô RNN như được mô tả trong Hình (2)

    Các tham số:
    xt -- dữ liệu đầu vào tại bước thời gian "t", mảng numpy có kích thước (n_x, m).
    a_prev -- trạng thái ẩn tại bước thời gian "t-1", mảng numpy có kích thước (n_a, m)
    parameters -- từ điển (dictionary) Python chứa:
        Wax -- Ma trận trọng số nhân với đầu vào, mảng numpy có kích thước (n_a, n_x)
        Waa -- Ma trận trọng số nhân với trạng thái ẩn, mảng numpy có kích thước (n_a, n_a)
        Wya -- Ma trận trọng số liên kết trạng thái ẩn với đầu ra, mảng numpy có kích thước (n_y, n_a)
        ba --  Hệ số bias, mảng numpy có kích thước (n_a, 1)
        by -- Hệ số bias liên kết trạng thái ẩn với đầu ra, mảng numpy có kích thước (n_y, 1)
    
    Giá trị trả về:
    a_next -- trạng thái ẩn tiếp theo, có kích thước (n_a, m)
    yt_pred -- dự đoán tại bước thời gian "t", mảng numpy có kích thước (n_y, m)
    cache -- bộ giá trị (tuple) cần thiết cho quá trình lan truyền ngược (backward pass), chứa (a_next, a_prev, xt, parameters)
    """
    
    ## Lấy các tham số từ "parameters"
    Wax = parameters['Wax']
    Waa = parameters['Waa']
    Wya = parameters['Wya']
    ba = parameters['ba']
    by = parameters['by']
    ### BẮT ĐẦU MÃ TẠI ĐÂY ### (≈2 dòng)
    # tính toán trạng thái kích hoạt tiếp theo sử dụng công thức đã cho ở trên
    a_next = np.tanh(np.dot(Waa , a_prev)+np.dot(Wax , xt)+ba)
    
    # tính toán đầu ra của ô hiện tại bằng công thức đã cho ở trên
    yt_pred = softmax(np.dot(Wya , a_next)+by)
    
    # lưu các giá trị cần thiết cho quá trình lan truyền ngược vào bộ nhớ đệm (cache)
    cache = (a_next , a_prev , xt , parameters)
    
    return a_next , yt_pred , cache
    
````

````python
#np.random.randn() là một hàm trong thư viện NumPy của Python dùng để khởi tạo một hoặc nhiều số ngẫu nhiên theo phân phối chuẩn chuẩn hóa (Standard Normal Distribution).
np.random.seed(1)
xt_tmp = np.random.randn(3,10)# 3 đặc trưng đầu vào (input features), 10 mẫu
a_prev_tmp = np.random.randn(5,10) # 5 hidden units, 10 mẫu
parameters_tmp = {}
parameters_tmp['Waa'] = np.random.randn(5,5)
parameters_tmp['Wax'] = np.random.randn(5, 3)
parameters_tmp['Wya'] = np.random.randn(2, 5)
parameters_tmp['ba'] = np.random.randn(5, 1)
parameters_tmp['by'] = np.random.randn(2, 1)
a_next_tmp, yt_pred_tmp, cache_tmp = rnn_cell_forward(xt_tmp, a_prev_tmp, parameters_tmp)
print("a_next[4] = \n", a_next_tmp[4])
print("a_next.shape = \n", a_next_tmp.shape)
print("yt_pred[1] =\n", yt_pred_tmp[1])
print("yt_pred.shape = \n", yt_pred_tmp.shape)

# UNIT TESTS
rnn_cell_forward_tests(rnn_cell_forward)
````

### 1.2 - Lan truyền xuôi (Forward Pass) trong RNN

- Mạng thần kinh tái phát (RNN) là sự lặp lại của ô RNN mà bạn vừa xây dựng. 
- Nếu chuỗi dữ liệu đầu vào của bạn có độ dài 10 bước thời gian (time steps), thì bạn sẽ tái sử dụng ô RNN đó 10 lần.
- Mỗi ô nhận hai đầu vào tại mỗi bước thời gian:
- $a^{\langle t-1 \rangle}$: Trạng thái ẩn (hidden state) từ ô trước đó.
- $x^{\langle t \rangle}$: Dữ liệu đầu vào tại bước thời gian hiện tại.
- Nó tạo ra hai đầu ra tại mỗi bước thời gian:
- Một trạng thái ẩn ($a^{\langle t \rangle}$)
- Một dự đoán ($y^{\langle t \rangle}$)
- Các trọng số và hệ số điều chỉnh (bias) $(W_{aa}, b_{a}, W_{ax}, b_{x})$ được tái sử dụng ở mỗi bước thời gian.
- Chúng được lưu giữ giữa các lần gọi hàm `rnn_cell_forward` trong từ điển 'parameters'.

Mạng xuất ra kết quả $y = (y^{\langle 1 \rangle}, y^{\langle 2 \rangle}, ..., y^{\langle T_x \rangle})$. </center></caption>

### Bài tập 2 - rnn_forward

Triển khai quá trình lan truyền tiến (forward propagation) của RNN được mô tả trong Hình 3.

**Hướng dẫn**:
* Tạo một mảng 3 chiều gồm các số không, $a$, có kích thước $(n_{a}, m, T_{x})$ để lưu trữ tất cả các trạng thái ẩn được tính toán bởi RNN
* Tạo một mảng 3 chiều gồm các số không, $\hat{y}$, có kích thước $(n_{y}, m, T_{x})$ để lưu trữ các dự đoán
- Lưu ý rằng trong trường hợp này, $T_{y} = T_{x}$ (dự đoán và đầu vào có cùng số lượng bước thời gian)
* Khởi tạo trạng thái ẩn 2 chiều `a_next` bằng cách gán nó bằng trạng thái ẩn ban đầu, $a_{0}$
* Tại mỗi bước thời gian $t$:
- Lấy $x^{\langle t \rangle}$, là một lát cắt 2 chiều của $x$ tại một bước thời gian $t$ cụ thể
- $x^{\langle t \rangle}$ có kích thước $(n_{x}, m)$
- $x$ có kích thước $(n_{x}, m, T_{x})$
- Cập nhật trạng thái ẩn 2 chiều $a^{\langle t \rangle}$ (tên biến là `a_next`), dự đoán $\hat{y}^{\langle t \rangle}$ và bộ nhớ đệm (cache) bằng cách chạy hàm `rnn_cell_forward`
- $a^{\langle t \rangle}$ có kích thước $(n_{a}, m)$
- Lưu trạng thái ẩn 2 chiều vào tensor 3 chiều $a$, tại vị trí thứ $t$
- $a$ có kích thước $(n_{a}, m, T_{x})$
- Lưu dự đoán $\hat{y}^{\langle t \rangle}$ 2 chiều (tên biến là `yt_pred`) vào tensor 3 chiều $\hat{y}_{pred}$ tại vị trí thứ $t$
- $\hat{y}^{\langle t \rangle}$ có kích thước $(n_{y}, m)$
- $\hat{y}$ có kích thước $(n_{y}, m, T_x)$
- Thêm cache vào danh sách các cache
* Trả về tensor 3 chiều $a$ và $\hat{y}$, cũng như danh sách các cache

#### Gợi ý bổ sung
- Một số tài liệu hữu ích về [np.zeros](https://docs.scipy.org/doc/numpy/reference/generated/numpy.zeros.html)
- Nếu bạn có một mảng numpy 3 chiều và muốn truy cập theo chiều thứ ba, bạn có thể sử dụng kỹ thuật cắt mảng (slicing) như sau: `var_name[:,:,i]`

````python
def rnn_forward(x, a0, parameters):
    """
    Triển khai quá trình lan truyền tiến (forward propagation) của mạng thần kinh tái phát (RNN) được mô tả trong Hình (3).

    Các tham số đầu vào:
    x -- Dữ liệu đầu vào cho mỗi bước thời gian, có kích thước (n_x, m, T_x).
    Với n_x: Số đặc trưng đầu vào tại mỗi thời điểm , m: Số mẫu trong batch và
    T_x là Time-step,  một vị trí trong chuỗi dữ liệu. Đơn vị của nó phụ thuộc vào cách lấy mẫu, ví dụ dữ liệu thời tiết mỗi giờ thì 1 time-step = 1 giờ. 
    
    a0 -- Trạng thái ẩn ban đầu, có kích thước (n_a, m) , với n_a: Số neuron của hidden state.
    
    parameters -- Từ điển Python chứa trọng số và bias của RNN :
    Waa -- Ma trận trọng số nhân với trạng thái ẩn, mảng numpy có kích thước (n_a, n_a)
    Wax -- Ma trận trọng số nhân với dữ liệu đầu vào, mảng numpy có kích thước (n_a, n_x)
    Wya -- Ma trận trọng số liên kết trạng thái ẩn với đầu ra, mảng numpy có kích thước (n_y, n_a)
    ba -- Mảng numpy chứa giá trị bias (độ lệch), có kích thước (n_a, 1)
    by -- Mảng numpy chứa giá trị bias liên kết trạng thái ẩn với đầu ra, có kích thước (n_y, 1)


    Giá trị trả về:
    a -- Các trạng thái ẩn tại mỗi bước thời gian, mảng numpy có kích thước (n_a, m, T_x)
    y_pred -- Các giá trị dự đoán tại mỗi bước thời gian, mảng numpy có kích thước (n_y, m, T_x)
    caches -- Bộ giá trị (tuple) cần thiết cho quá trình lan truyền ngược (backward pass), bao gồm (danh sách các giá trị cache, x)
    """
    # Khởi tạo "caches" để chứa danh sách tất cả các cache
    caches = [] #Danh sách này lưu thông tin trung gian của mỗi time-step để phục vụ Backward Propagation Through Time (BPTT) sau này.

    # Lấy các kích thước dữ liệu từ hình dạng của x và parameters["Wya"]
    n_x , m , T_x = x.shape
    n_y , n_a = parameters['Wya'].shape # RNN có 4 hidden units và dự đoán 2 giá trị đầu ra tại mỗi time-step
    
    #Khởi tạo  tensor lưu kết quả ("a" và "y_pred" bằng các giá trị không)
    a = np.zeros((n_a , m , T_x))
    y_pred = np.zeros((n_y , m , T_x))
    #Hai tensor này dùng để lưu kết quả của tất cả các time-step, không chỉ time-step cuối cùng.
    
    
    # Khởi tạo hidden state.Ban đầu chưa xử lý time-step nào, nên sử dụng hidden state ban đầu a0.
    a_next = a0
    
    #Duyệt toàn bộ time-step
    for t in range(T_x) :
        
        #x[:, :, t]: Dữ liệu của toàn bộ m mẫu tại time-step t.a_next: Hidden state từ bước trước.- parameters: Trọng số và bias.
        a_next , yt_pred , cache = rnn_cell_forward((x[:,:,t]) , a_next , parameters)
        
        # Lưu giá trị của trạng thái ẩn "tiếp theo" mới vào một biến (≈1 dòng)
        a[:,:,t]= a_next 
        
        #Save the value of the prediction in y (≈1 line)
        y_pred[:,:,t]= yt_pred
        
        caches.append(cache)
    
    #Đóng gói danh sách cache và toàn bộ dữ liệu x thành một tuple, phục vụ BPTT.
    caches = (caches, x)
    
    return a , y_pred , caches
    
    
````

````python
np.random.seed(1)
x_tmp = np.random.randn(3,10,4)
a0_tmp = np.random.randn(5,10)
parameters_tmp = {}
parameters_tmp['Waa'] = np.random.randn(5,5)
parameters_tmp['Wax'] = np.random.randn(5, 3)
parameters_tmp['Wya'] = np.random.randn(2, 5)
parameters_tmp['ba'] = np.random.randn(5, 1)
parameters_tmp['by'] = np.random.randn(2, 1)

a_tmp , y_pred_tmp , caches_tmp = rnn_forward(x_tmp , a0_tmp , parameters_tmp)
print("a[4][1] = \n", a_tmp[4][1])
print("a.shape = \n", a_tmp.shape)
print("y_pred[1][3] =\n", y_pred_tmp[1][3])
print("y_pred.shape = \n", y_pred_tmp.shape)
print("caches[1][1][3] =\n", caches_tmp[1][1][3])
print("len(caches) = \n", len(caches_tmp))

#UNIT TEST    
rnn_forward_test(rnn_forward)
````

* Mỗi ô tạo ra hai đầu ra tại mỗi bước thời gian:
* Một trạng thái ẩn

Hình dưới đây minh họa các hoạt động của một ô LSTM:

<caption><center><font color='purple'><b>Hình 4</b>: Ô LSTM. Ô này theo dõi và cập nhật "trạng thái ô" (cell state) hay biến bộ nhớ $c^{\langle t \rangle}$ tại mỗi bước thời gian; trạng thái này có thể khác với $a^{\langle t \rangle}$.
Lưu ý: thành phần $softmax^{}$ bao gồm một lớp kết nối đầy đủ (dense layer) và hàm softmax.</center></caption>

Tương tự như ví dụ về RNN ở trên, bạn sẽ bắt đầu bằng việc triển khai ô LSTM cho một bước thời gian duy nhất. Sau đó, bạn sẽ gọi ô này lặp lại nhiều lần bên trong một vòng lặp "for" để xử lý dữ liệu đầu vào có $T_x$ bước thời gian.

### Tổng quan về các cổng và trạng thái

#### Cổng quên (forget gate) $\mathbf{\Gamma}_{f}$

* Giả sử bạn đang đọc các từ trong một văn bản và dự định sử dụng LSTM để theo dõi các cấu trúc ngữ pháp, chẳng hạn như việc chủ ngữ là số ít ("puppy" - chó con) hay số nhiều ("puppies" - những chú chó con).
* Nếu trạng thái của chủ ngữ thay đổi (từ từ số ít sang từ số nhiều), thông tin về trạng thái trước đó sẽ trở nên lỗi thời, do đó bạn sẽ "quên" trạng thái lỗi thời đó đi.
* "Cổng quên" là một tensor chứa các giá trị nằm trong khoảng từ 0 đến 1.
* Nếu một đơn vị trong cổng quên có giá trị gần bằng 0, LSTM sẽ "quên" trạng thái được lưu trữ tại đơn vị tương ứng của trạng thái ô (cell state) trước đó. 
* Nếu một đơn vị trong cổng quên có giá trị gần bằng 1, LSTM sẽ giữ lại (nhớ) phần lớn giá trị tương ứng trong trạng thái đã lưu.

##### Phương trình

$$\mathbf{\Gamma}_f^{\langle t \rangle} = \sigma(\mathbf{W}_f[\mathbf{a}^{\langle t-1 \rangle}, \mathbf{x}^{\langle t \rangle}] + \mathbf{b}_f)\tag{1} $$

##### Giải thích phương trình:

* $\mathbf{W_{f}}$ chứa các trọng số chi phối hoạt động của cổng quên.
* Trạng thái ẩn của bước thời gian trước đó $[\mathbf{a}^{\langle t-1 \rangle}]$ và đầu vào của bước thời gian hiện tại $[\mathbf{x}^{\langle t \rangle}]$ được ghép lại với nhau và nhân với $\mathbf{W_{f}}$.
* Hàm sigmoid được sử dụng để đưa các giá trị của tensor cổng $\mathbf{\Gamma}_f^{\langle t \rangle}$ về khoảng từ 0 đến 1.
* Cổng quên $\mathbf{\Gamma}_f^{\langle t \rangle}$ có cùng kích thước với trạng thái ô trước đó $\mathbf{c}^{\langle t-1 \rangle}$.
* Điều này có nghĩa là hai đại lượng này có thể được nhân với nhau theo từng phần tử (element-wise). * Việc nhân các tensor $\mathbf{\Gamma}_f^{\langle t \rangle} * \mathbf{c}^{\langle t-1 \rangle}$ giống như áp dụng một lớp mặt nạ (mask) lên trạng thái ô (cell state) trước đó.
* Nếu một giá trị trong $\mathbf{\Gamma}_f^{\langle t \rangle}$ bằng 0 hoặc gần bằng 0, thì tích thu được sẽ gần bằng 0.
* Điều này ngăn chặn việc lưu giữ thông tin tại đơn vị tương ứng trong $\mathbf{c}^{\langle t-1 \rangle}$ cho bước thời gian tiếp theo.
* Tương tự, nếu một giá trị gần bằng 1, tích thu được sẽ gần bằng giá trị gốc trong trạng thái ô trước đó. 
* LSTM sẽ giữ lại thông tin từ đơn vị tương ứng của $\mathbf{c}^{\langle t-1 \rangle}$ để sử dụng cho bước thời gian tiếp theo.

##### Tên biến trong mã nguồn
Tên các biến trong mã nguồn tương tự như trong các phương trình, nhưng có một số khác biệt nhỏ.
* `Wf`: trọng số của cổng quên (forget gate) $\mathbf{W}_{f}$
* `bf`: hệ số điều chỉnh (bias) của cổng quên $\mathbf{b}_{f}$
* `ft`: cổng quên $\Gamma_f^{\langle t \rangle}$

#### Giá trị ứng viên $\tilde{\mathbf{c}}^{\langle t \rangle}$
* Giá trị ứng viên là một tensor chứa thông tin từ bước thời gian hiện tại, thông tin này **có thể** được lưu trữ trong trạng thái ô hiện tại $\mathbf{c}^{\langle t \rangle}$.
* Các phần của giá trị ứng viên được truyền đi sẽ phụ thuộc vào cổng cập nhật (update gate).
* Giá trị ứng viên là một tensor chứa các giá trị nằm trong khoảng từ -1 đến 1.
* Ký hiệu dấu ngã "~" được sử dụng để phân biệt giá trị ứng viên $\tilde{\mathbf{c}}^{\langle t \rangle}$ với trạng thái ô $\mathbf{c}^{\langle t \rangle}$.

##### Phương trình
$$\mathbf{\tilde{c}}^{\langle t \rangle} = \tanh\left( \mathbf{W}_{c} [\mathbf{a}^{\langle t - 1 \rangle}, \mathbf{x}^{\langle t \rangle}] + \mathbf{b}_{c} \right) \tag{3}$$

##### Giải thích phương trình
* Hàm *tanh* tạo ra các giá trị nằm trong khoảng từ -1 đến 1.


##### Tên biến trong mã nguồn
* `cct`: giá trị ứng viên $\mathbf{\tilde{c}}^{\langle t \rangle}$

#### Cổng cập nhật $\mathbf{\Gamma}_{i}$

* Bạn sử dụng cổng cập nhật để quyết định những khía cạnh nào của giá trị ứng viên $\tilde{\mathbf{c}}^{\langle t \rangle}$ sẽ được thêm vào trạng thái ô nhớ $c^{\langle t \rangle}$.
* Cổng cập nhật quyết định phần nào của tensor "ứng viên" $\tilde{\mathbf{c}}^{\langle t \rangle}$ được truyền sang trạng thái ô nhớ $\mathbf{c}^{\langle t \rangle}$.
* Cổng cập nhật là một tensor chứa các giá trị nằm trong khoảng từ 0 đến 1.
* Khi một đơn vị trong cổng cập nhật có giá trị gần bằng 1, nó cho phép giá trị của ứng viên $\tilde{\mathbf{c}}^{\langle t \rangle}$ được truyền sang trạng thái ô nhớ $\mathbf{c}^{\langle t \rangle}$.
* Khi một đơn vị trong cổng cập nhật có giá trị gần bằng 0, nó ngăn chặn giá trị tương ứng trong ứng viên được truyền sang trạng thái ô nhớ.
* Lưu ý rằng chỉ số dưới "i" được sử dụng thay vì "u", nhằm tuân theo quy ước trong các tài liệu chuyên ngành.

##### Phương trình

$$\mathbf{\Gamma}_i^{\langle t \rangle} = \sigma(\mathbf{W}_i[a^{\langle t-1 \rangle}, \mathbf{x}^{\langle t \rangle}] + \mathbf{b}_i)\tag{2} $$

##### Giải thích phương trình

* Tương tự như cổng quên, tại đây hàm sigmoid tạo ra các giá trị nằm trong khoảng từ 0 đến 1 cho $\mathbf{\Gamma}_i^{\langle t \rangle}$.
* Cổng cập nhật được nhân theo từng phần tử (element-wise) với giá trị ứng viên, và tích số này ($\mathbf{\Gamma}_{i}^{\langle t \rangle} * \tilde{c}^{\langle t \rangle}$) được sử dụng để xác định trạng thái ô nhớ $\mathbf{c}^{\langle t \rangle}$.

##### Tên biến trong mã nguồn (Lưu ý rằng chúng khác với các ký hiệu trong phương trình)
Trong mã nguồn, bạn sẽ sử dụng các tên biến thường thấy trong tài liệu học thuật. Các biến này không sử dụng ký tự "u" để biểu thị cho "update" (cập nhật). * `Wi` là trọng số của cổng cập nhật $\mathbf{W}_i$ (không phải "Wu")
* `bi` là hệ số điều chỉnh (bias) của cổng cập nhật $\mathbf{b}_i$ (không phải "bu")
* `it` là cổng cập nhật $\mathbf{\Gamma}_i^{\langle t \rangle}$ (không phải "ut")

#### Trạng thái ô $\mathbf{c}^{\langle t \rangle}$

* Trạng thái ô là "bộ nhớ" được truyền sang các bước thời gian tiếp theo.
* Trạng thái ô mới $\mathbf{c}^{\langle t \rangle}$ là sự kết hợp giữa trạng thái ô trước đó và giá trị ứng viên.

##### Phương trình

$$ \mathbf{c}^{\langle t \rangle} = \mathbf{\Gamma}_f^{\langle t \rangle}* \mathbf{c}^{\langle t-1 \rangle} + \mathbf{\Gamma}_{i}^{\langle t \rangle} *\mathbf{\tilde{c}}^{\langle t \rangle} \tag{4} $$

##### Giải thích phương trình
* Trạng thái ô trước đó $\mathbf{c}^{\langle t-1 \rangle}$ được điều chỉnh (gán trọng số) bởi cổng quên $\mathbf{\Gamma}_{f}^{\langle t \rangle}$
* và giá trị ứng viên $\tilde{\mathbf{c}}^{\langle t \rangle}$ được điều chỉnh (gán trọng số) bởi cổng cập nhật $\mathbf{\Gamma}_{i}^{\langle t \rangle}$

##### Tên biến và kích thước trong mã nguồn
* `c`: trạng thái ô, bao gồm tất cả các bước thời gian, $\mathbf{c}$ có kích thước $(n_{a}, m, T_x)$
* `c_next`: trạng thái ô mới (tiếp theo), $\mathbf{c}^{\langle t \rangle}$ có kích thước $(n_{a}, m)$
* `c_prev`: trạng thái ô trước đó, $\mathbf{c}^{\langle t-1 \rangle}$, có kích thước $(n_{a}, m)$

#### Cổng đầu ra $\mathbf{\Gamma}_{o}$

* Cổng đầu ra quyết định nội dung nào sẽ được gửi đi dưới dạng dự đoán (đầu ra) của bước thời gian.
* Giống như các cổng khác, cổng đầu ra chứa các giá trị nằm trong khoảng từ 0 đến 1.

##### Phương trình

$$ \mathbf{\Gamma}_o^{\langle t \rangle}=  \sigma(\mathbf{W}_o[\mathbf{a}^{\langle t-1 \rangle}, \mathbf{x}^{\langle t \rangle}] + \mathbf{b}_{o})\tag{5}$$

##### Giải thích phương trình
* Cổng đầu ra được xác định bởi trạng thái ẩn trước đó $\mathbf{a}^{\langle t-1 \rangle}$ và đầu vào hiện tại $\mathbf{x}^{\langle t \rangle}$.
* Hàm sigmoid đưa giá trị của cổng về khoảng từ 0 đến 1.


##### Tên biến trong mã nguồn
* `Wo`: trọng số cổng đầu ra, $\mathbf{W_o}$
* `bo`: hệ số điều chỉnh (bias) cổng đầu ra, $\mathbf{b_o}$
* `ot`: cổng đầu ra, $\mathbf{\Gamma}_{o}^{\langle t \rangle}$

#### Trạng thái ẩn $\mathbf{a}^{\langle t \rangle}$

* Trạng thái ẩn được truyền sang bước thời gian tiếp theo của ô LSTM.
* Nó được sử dụng để xác định ba cổng ($\mathbf{\Gamma}_{f}, \mathbf{\Gamma}_{u}, \mathbf{\Gamma}_{o}$) của bước thời gian kế tiếp.
* Trạng thái ẩn cũng được sử dụng để đưa ra dự đoán $y^{\langle t \rangle}$.

##### Phương trình

$$ \mathbf{a}^{\langle t \rangle} = \mathbf{\Gamma}_o^{\langle t \rangle} * \tanh(\mathbf{c}^{\langle t \rangle})\tag{6} $$

##### Giải thích phương trình
* Trạng thái ẩn $\mathbf{a}^{\langle t \rangle}$ được xác định bởi trạng thái ô $\mathbf{c}^{\langle t \rangle}$ kết hợp với cổng đầu ra $\mathbf{\Gamma}_{o}$.
* Trạng thái ô được đưa qua hàm `tanh` để đưa các giá trị về khoảng từ -1 đến 1.
* Cổng đầu ra hoạt động như một "mặt nạ" (mask), quyết định việc giữ lại các giá trị của $\tanh(\mathbf{c}^{\langle t \rangle})$ hay ngăn không cho các giá trị đó được đưa vào trạng thái ẩn $\mathbf{a}^{\langle t \rangle}$.

##### Tên biến và kích thước trong mã nguồn
* `a`: trạng thái ẩn, bao gồm các bước thời gian. $\mathbf{a}$ có kích thước $(n_{a}, m, T_{x})$
* `a_prev`: trạng thái ẩn từ bước thời gian trước đó. $\mathbf{a}^{\langle t-1 \rangle}$ có kích thước $(n_{a}, m)$
* `a_next`: trạng thái ẩn cho bước thời gian tiếp theo. $\mathbf{a}^{\langle t \rangle}$ có kích thước $(n_{a}, m)$

#### Dự đoán $\mathbf{y}^{\langle t \rangle}_{pred}$
* Dự đoán trong trường hợp sử dụng này là bài toán phân loại, do đó bạn sẽ sử dụng hàm softmax.

Công thức là:
$$\mathbf{y}^{\langle t \rangle}_{pred} = \textrm{softmax}(\mathbf{W}_{y} \mathbf{a}^{\langle t \rangle} + \mathbf{b}_{y})$$

##### Tên biến và kích thước trong mã nguồn
* `y_pred`: kết quả dự đoán, bao gồm tất cả các bước thời gian. $\mathbf{y}_{pred}$ có kích thước $(n_{y}, m, T_{x})$. Lưu ý rằng trong ví dụ này, $(T_{y} = T_{x})$.
* `yt_pred`: kết quả dự đoán cho bước thời gian hiện tại $t$. $\mathbf{y}^{\langle t \rangle}_{pred}$ có kích thước $(n_{y}, m)$.
