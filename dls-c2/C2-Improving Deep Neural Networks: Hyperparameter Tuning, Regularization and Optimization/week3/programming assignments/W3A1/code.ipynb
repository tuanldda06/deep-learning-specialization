# Giới thiệu về TensorFlow

Chào mừng bạn đến với bài tập lập trình của tuần này! Cho đến nay, bạn vẫn luôn sử dụng Numpy để xây dựng các mạng nơ-ron, nhưng tuần này bạn sẽ khám phá một framework học sâu (deep learning framework) giúp việc xây dựng mạng nơ-ron trở nên dễ dàng hơn. Các framework học máy như TensorFlow, PaddlePaddle, Torch, Caffe, Keras và nhiều framework khác có thể đẩy nhanh đáng kể quá trình phát triển các mô hình học máy của bạn. TensorFlow 2.3 đã có những cải tiến vượt bậc so với phiên bản tiền nhiệm, và bạn sẽ được trải nghiệm cũng như triển khai một số cải tiến đó ngay tại đây!

Sau khi hoàn thành bài tập này, bạn sẽ có thể thực hiện các thao tác sau trong TensorFlow 2.3:

* Sử dụng `tf.Variable` để thay đổi trạng thái của một biến
* Giải thích sự khác biệt giữa biến (variable) và hằng số (constant)
* Huấn luyện một mạng nơ-ron trên tập dữ liệu TensorFlow

Các framework lập trình như TensorFlow không chỉ giúp tiết kiệm thời gian viết mã mà còn có khả năng thực hiện các tối ưu hóa giúp mã nguồn chạy nhanh hơn.

## Lưu ý quan trọng về việc nộp bài cho hệ thống chấm điểm tự động (AutoGrader)

Trước khi nộp bài tập cho AutoGrader, vui lòng đảm bảo bạn không thực hiện các hành động sau:

1. Thêm bất kỳ câu lệnh `print` _nào khác_ vào bài tập.
2. Thêm bất kỳ ô mã (code cell) _nào khác_ vào bài tập.
3. Thay đổi bất kỳ tham số nào của hàm.
4. Sử dụng biến toàn cục (global variable) bên trong các bài tập được chấm điểm. Trừ khi có hướng dẫn cụ thể, vui lòng tránh sử dụng biến toàn cục và thay vào đó hãy sử dụng biến cục bộ (local variable).
5. Thay đổi mã nguồn của bài tập ở những phần không yêu cầu, chẳng hạn như tạo thêm các biến _không cần thiết_.

Nếu thực hiện bất kỳ hành động nào kể trên, bạn có thể gặp lỗi như `Grader not found` (hoặc các lỗi bất thường tương tự) khi nộp bài. Hãy kiểm tra kỹ các vấn đề này trước khi tìm kiếm sự trợ giúp hoặc gỡ lỗi (debug) cho bài tập của mình. Nếu đúng như vậy và bạn không nhớ những thay đổi mình đã thực hiện, bạn có thể lấy một bản sao mới của bài tập bằng cách làm theo các [hướng dẫn](https://www.coursera.org/learn/deep-neural-network/supplement/QWEnZ/h-ow-to-refresh-your-workspace) này.
