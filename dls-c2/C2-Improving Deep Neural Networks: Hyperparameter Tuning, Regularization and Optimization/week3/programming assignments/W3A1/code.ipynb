## 2 - Tối ưu hóa cơ bản với GradientTape

Điểm tuyệt vời của TensorFlow 2 nằm ở sự đơn giản. Về cơ bản, tất cả những gì bạn cần làm là triển khai quá trình lan truyền tiến (forward propagation) thông qua một đồ thị tính toán. TensorFlow sẽ tự động tính toán các đạo hàm giúp bạn bằng cách đi ngược lại đồ thị đã được ghi lại bởi `GradientTape`. Sau đó, việc còn lại chỉ là xác định hàm chi phí (cost function) và thuật toán tối ưu hóa (optimizer) mà bạn muốn sử dụng!

Khi viết chương trình TensorFlow, đối tượng chính được sử dụng và biến đổi là `tf.Tensor`. Các tensor này tương đương với mảng Numpy trong TensorFlow; nghĩa là chúng là các mảng đa chiều với kiểu dữ liệu xác định, đồng thời chứa thông tin về đồ thị tính toán.

Dưới đây, bạn sẽ sử dụng `tf.Variable` để lưu trữ trạng thái các biến của mình. Các biến chỉ có thể được tạo một lần vì giá trị khởi tạo của chúng sẽ xác định hình dạng (shape) và kiểu dữ liệu (type) của biến đó. Ngoài ra, tham số `dtype` trong `tf.Variable` có thể được thiết lập để cho phép chuyển đổi dữ liệu sang kiểu đó. Tuy nhiên, nếu không được chỉ định cụ thể, kiểu dữ liệu sẽ được giữ nguyên (nếu giá trị khởi tạo là một Tensor) hoặc được quyết định bởi hàm `convert_to_tensor`. Tốt nhất là bạn nên chỉ định trực tiếp để tránh xảy ra lỗi!
